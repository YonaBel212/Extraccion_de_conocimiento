```python
import pandas as pd
import random
```

```python
datos_base = {
    "influencer": ["@cancun_travel", "@playa_del_carmen", "@mexicobeach", 
                   "@lifestylemx", "@cancun_hotels", "@viajerosmx"],
    "likes_base": [5000, 3000, 8000, 2000, 4500, 6000],
    "comentarios_base": [200, 150, 300, 80, 180, 250]
}
dfBase = pd.DataFrame(datos_base)
print("📌 Datos base de influencers:")
print(dfBase)
```

```python
cantidad = 1000000
tipos_post = ["foto", "reel", "carrusel", "historia"]

dfInstagram = pd.DataFrame({
    "influencer": [random.choice(dfBase["influencer"]) for _ in range(cantidad)],
    "likes": [random.randint(100, 15000) for _ in range(cantidad)],
    "comentarios": [random.randint(0, 500) for _ in range(cantidad)],
    "tipo_post": [random.choice(tipos_post) for _ in range(cantidad)],
    "fecha": [f"2025-{random.randint(1,12):02d}-{random.randint(1,28):02d}" for _ in range(cantidad)]
})

print(f"\nDataset creado con {dfInstagram.shape[0]:,} publicaciones")
print(f"Columnas: {dfInstagram.shape[1]}")
```

```python
print("\n📋 Primeras 10 publicaciones:")
print(dfInstagram.head(10))
```

```python
dfInstagram["engagement"] = dfInstagram["likes"] + dfInstagram["comentarios"]
print("\n Estadísticas generales:")
print(dfInstagram[["likes", "comentarios", "engagement"]].describe())
```

```python
print("\n Top influencers por engagement total:")
engagement_por_influencer = dfInstagram.groupby("influencer")["engagement"].sum().sort_values(ascending=False)
print(engagement_por_influencer)
```

```python
print("\n Rendimiento por tipo de post:")
rendimiento_tipo = dfInstagram.groupby("tipo_post")[["likes", "comentarios", "engagement"]].mean()
print(rendimiento_tipo.round(2))
```

```python
dfInstagram.to_csv("instagram_dataset_1M.csv", index=False)
print("\n Archivo guardado: instagram_dataset_1M.csv")
```
